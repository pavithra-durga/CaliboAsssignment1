```python
import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "DBMS", "DSA", "CN", "AI"]

yash = [78, 85, 72, 90, 81, 88]
kushi = [85, 79, 88, 84, 76, 92]

fig, ax = plt.subplots(figsize=(8, 4.5))

ax.plot(subjects, yash, marker="o", color="#065A82", label="Yash")
ax.plot(subjects, kushi, marker="s", color="#D95F02", label="Kushi")

ax.set_title("Yash vs Kushi - Subject-wise Marks")
ax.set_xlabel("Subjects")
ax.set_ylabel("Marks")
ax.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

subjects = ["Python", "SQL", "DBMS", "DSA", "CN", "AI"]

yash = [78, 85, 72, 90, 81, 88]
kushi = [85, 79, 88, 84, 76, 92]

x = np.arange(len(subjects))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 4.5))

ax.bar(x - width/2, yash, width, color="#065A82", label="Yash")
ax.bar(x + width/2, kushi, width, color="#D95F02", label="Kushi")

ax.set_title("Comparison of Yash and Kushi Marks")
ax.set_xlabel("Subjects")
ax.set_ylabel("Marks")
ax.set_xticks(x)
ax.set_xticklabels(subjects)
ax.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

yash = [78, 85, 72, 90, 81, 88]
kushi = [85, 79, 88, 84, 76, 92]

fig, ax = plt.subplots(figsize=(8, 4.5))

ax.hist(yash, bins=5, alpha=0.7, color="#065A82",
        edgecolor="black", label="Yash")

ax.hist(kushi, bins=5, alpha=0.7, color="#D95F02",
        edgecolor="black", label="Kushi")

ax.set_title("Distribution of Marks")
ax.set_xlabel("Marks")
ax.set_ylabel("Number of Subjects")
ax.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "DBMS", "DSA", "CN", "AI"]
yash = [78, 85, 72, 90, 81, 88]

fig, ax = plt.subplots(figsize=(7, 5))

ax.pie(
    yash,
    labels=subjects,
    autopct="%1.1f%%",
    startangle=90,
    colors=[
        "#065A82",
        "#D95F02",
        "#2A9D8F",
        "#E9C46A",
        "#E76F51",
        "#6A4C93"
    ]
)

ax.set_title("Yash's Marks Distribution by Subject")

plt.show()
```
