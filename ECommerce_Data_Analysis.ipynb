```python
# Numerical columns → median
df['Age'] = df['Age'].fillna(df['Age'].median())

df['Discount'] = df['Discount'].fillna(df['Discount'].median())

df['Rating'] = df['Rating'].fillna(df['Rating'].median())

df['Delivery_Days'] = df['Delivery_Days'].fillna(df['Delivery_Days'].median())


# Categorical columns → mode
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

df['City'] = df['City'].fillna(df['City'].mode()[0])

df['Payment_Method'] = df['Payment_Method'].fillna(df['Payment_Method'].mode()[0])
```

```python
#total missing values after treatment
print("Total missing values:", df.isnull().sum().sum())
```

### Missing Value Treatment - Interpretation

Missing values were identified in Age, Gender, City, Discount, Payment_Method, Rating, and Delivery_Days.

For numerical columns such as Age, Discount, Rating, and Delivery_Days, the median was used because it is less affected by extreme values and provides a representative central value.

For categorical columns such as Gender, City, and Payment_Method, the mode was used because it represents the most frequently occurring category.

The missing values were replaced instead of removing complete records, because deleting rows would unnecessarily reduce the dataset size and potentially remove useful information from other columns.

After applying the treatment, the dataset was checked again and contained zero missing values.

Task  9: Detect
Outliers Using IQR

1. Calculate Q1, Q3, IQR and bounds

```python
iqr_columns = ['Age', 'Unit_Price', 'Total_Amount']

for col in iqr_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    print(f"\n{'='*50}")
    print(f"Column: {col}")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
```

2. Identify and count outlier records

```python
for col in iqr_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[col] < lower_bound) | 
        (df[col] > upper_bound)
    ]
    
    print(f"\n{col}")
    print("Number of outliers:", len(outliers))
    display(outliers[[col]])
```

3. Better: Create an outlier summary table

```python
outlier_summary = []

for col in iqr_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_count = ((df[col] < lower_bound) | 
                     (df[col] > upper_bound)).sum()
    
    outlier_summary.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': outlier_count
    })

outlier_summary_df = pd.DataFrame(outlier_summary)

display(outlier_summary_df)
```

### IQR Outlier Analysis - Interpretation

The IQR method was used to identify statistical outliers in Age, Unit_Price, and Total_Amount. Values below the lower bound or above the upper bound were classified as potential outliers.

The presence of an outlier does not necessarily indicate a data-entry error. Extreme Age values may represent legitimate customers, while extreme Unit_Price and Total_Amount values may represent high-value products or transactions.

Therefore, the identified outliers will not be removed automatically. Unit_Price and Total_Amount outliers will be investigated further using order-amount consistency checks before making a final decision.

Expected Total=Unit Price×Quantity×(1−
Discount/100)
1. Calculate Expected Total

```python
df['Expected_Total'] = (
    df['Unit_Price'] *
    df['Quantity'] *
    (1 - df['Discount'] / 100)
)

df[['Unit_Price', 'Quantity', 'Discount',
    'Expected_Total', 'Total_Amount']].head()
```

```python
df['Difference'] = (
    df['Total_Amount'] - df['Expected_Total']
)

df[['Expected_Total', 'Total_Amount', 'Difference']].head()
```

records with large differences

```python
large_differences = df.loc[
    df['Difference'].abs().sort_values(ascending=False).index
]

display(
    large_differences[
        ['Order_ID', 'Product', 'Unit_Price',
         'Quantity', 'Discount',
         'Expected_Total', 'Total_Amount', 'Difference']
    ].head(20)
)
```

```python
df['Difference_Percentage'] = (
    abs(df['Difference']) / df['Expected_Total']
) * 100

display(
    df.sort_values(
        'Difference_Percentage',
        ascending=False
    )[
        ['Order_ID', 'Product', 'Unit_Price',
         'Quantity', 'Discount',
         'Expected_Total', 'Total_Amount',
         'Difference', 'Difference_Percentage']
    ].head(20)
)
```

```python
large_diff_records = df[
    df['Difference_Percentage'] > 10
]

print("Number of records with more than 10% difference:",
      len(large_diff_records))

display(
    large_diff_records[
        ['Order_ID', 'Product', 'Unit_Price',
         'Quantity', 'Discount',
         'Expected_Total', 'Total_Amount',
         'Difference', 'Difference_Percentage']
    ].head(20)
)
```
